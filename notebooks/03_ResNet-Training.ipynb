# ResNet Training

This notebook demonstrates how to train a ResNet model for motor fault detection.
We assume that:
- Data preprocessing has been completed and the processed data is available.
- We have a ResNet model defined in `src/models.py` and training functions defined in `src/train.py`.

```python
import torch
from src.models import ResNet, ResidualBlock
from src.data_preprocessing import create_datasets
from src.train import train_resnet_model

%load_ext autoreload
%autoreload 2
```

```python
# Set device
device: str = 'cuda' if torch.cuda.is_available() else ('mps' if torch.backends.mps.is_available() else 'cpu')
device = torch.device(device)
print(f"Using device: {device}")
```

```python
# Specify directories for normal and anomalous data
normal_data_dir = '../data/processed/Final_Normal'
anomalous_data_dir = '../data/processed/Final_Anomalous'
batch_size = 16

# Create the training and testing datasets
train_loader, test_loader = create_datasets(normal_data_dir, anomalous_data_dir, batch_size=batch_size)

print("Data loaders created.")
print(f"Number of training batches: {len(train_loader)}")
print(f"Number of testing batches: {len(test_loader)}")
```

```python
# Initialize the ResNet model with a basic configuration
model = ResNet(ResidualBlock, [2, 2, 2, 2], num_classes=2)
model.to(device)

print("Model initialized.")
```

```python
# Train the model
num_epochs = 5
train_resnet_model(model, train_loader, test_loader, device, num_epochs=num_epochs)

print("Training completed.")
```
